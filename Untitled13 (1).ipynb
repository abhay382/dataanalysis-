```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Netflix") \
    .getOrCreate()

print("Spark Working ✅")
```

```python
df = spark.read.csv("netflix_titles.csv", header=True, inferSchema=True)

df.show(5)
```

```python
df.printSchema()
```

```python
print("Total records:", df.count())
```

```python
from pyspark.sql.functions import col, to_date

df = df.dropna(subset=["type", "country", "release_year"])

df = df.withColumn(
    "date_added",
    to_date(col("date_added"), "MMMM d, yyyy")
)

df.show(5)
```

```python
df.groupBy("type").count().show()
```

```python
df.groupBy("country") \
  .count() \
  .orderBy("count", ascending=False) \
  .show(10)
```

```python
df.groupBy("release_year") \
  .count() \
  .orderBy("release_year") \
  .show()
```

```python
from pyspark.sql.functions import split, explode

genre_df = df.withColumn(
    "genre",
    explode(split(col("listed_in"), ", "))
)

genre_df.groupBy("genre") \
        .count() \
        .orderBy("count", ascending=False) \
        .show(10)
```

```python
pdf = df.toPandas()
```

```python
import matplotlib.pyplot as plt

pdf['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

```python
pdf['country'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Countries")
plt.show()
```
